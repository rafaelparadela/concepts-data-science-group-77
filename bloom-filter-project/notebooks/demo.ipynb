```python
import sys
import os
sys.path.append(os.path.abspath(".."))
from bloomfilter.bloom_filter import BloomFilter
bf = BloomFilter(size=10000, hash_count=5)

bf.add("apple")
bf.add("banana")

print(bf.contains("apple"))   # True
print(bf.contains("banana"))  # True
print(bf.contains("grape"))   # Probably False
```
